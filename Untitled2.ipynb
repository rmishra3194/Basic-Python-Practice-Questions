```python
#Q9 (max score from a list of (name, score) tuples using max() with a key)?

students = [("Alice", 85), ("Bob", 92), ("Charlie", 78), ("Diana", 92)]

print(max(students, key = lambda x : x[1]))
```

```python
#Q10: Given a dict {name: age}, use a dict comprehension to keep only entries where age > 18.

ages = {"Alice": 25, "Bob": 17, "Charlie": 19, "Diana": 15, "Eve": 30}

{x: ages[x] for x in ages if ages[x] > 18}
```

```python
#Q11 — merge two dicts using {**d1, **d2} and see what happens with duplicate keys?

d1 = {"a": 1, "b": 2, "c": 3}
d2 = {"b": 20, "c": 30, "d": 4}

merged = {**d1, **d2}
print(merged)

mer1 = d1 | d2
print(mer1)
```

```python
#Q12: Given a dict, use .items() and a comprehension to swap keys and values.

d = {"a": 1, "b": 2, "c": 3}

swap = {y : x for x, y in d.items()}
print(swap)
```

```python
#Q13: Given two lists, find common elements using set intersection (&).
list1 = [1, 2, 3, 4, 5, 6]
list2 = [4, 5, 6, 7, 8, 9]

common = set(list1) & set(list2)
print(common)
```

```python
#Q14: Given two lists, find elements in the first but not the second, using set difference (-).

list1 = [1, 2, 3, 4, 5, 6]
list2 = [4, 5, 6, 7, 8, 9]

diff = set(list1) - set(list2)
print(diff)
```

```python
#Q15: Check if a list has duplicates by comparing len(list) vs len(set(list)).

list1 = [1, 2, 3, 4, 5]        # no duplicates
list2 = [1, 2, 2, 3, 4, 4]     # has duplicates

duplicates = len(list2) - len(set(list2))
print(duplicates)
```

```python
#Q16: Generate squares of all odd numbers from 1–20 using a list comprehension.
#Data: just use range(1, 21) as your source range (1 to 20 inclusive).

sqr = [x*x for x in range(1, 21) if x%2!=0]
print(sqr)
```
