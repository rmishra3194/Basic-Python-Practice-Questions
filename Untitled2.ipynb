```python
#Check prime numbers

n = int(input("Enter a number: "))

if n <= 1:
  print("Not a prime number")
else:
  for i in range(2, int(n**0.5) + 1):
    if n % i == 0:
      print("Not a prime number")
      break
  else:
    print("Prime number")
```

```python
#Prime numbers in range

start = int(input("Start: "))
end = int(input("End: "))
print("Prime numbers in range:")

for num in range(start, end + 1):
    if num <= 1:
        continue
    for i in range(2, int(num**0.5) + 1):
        if num % i == 0:
            break
    else:
        print(num)
```

```python
#Find Factorial

n = int(input("Enter a number: "))
fact = 1

if n < 0:
    print("Factorial is not defined")
elif n == 0:
    print("Factorial = 1")
else:
    for i in range(1, n + 1):
        fact *= i
    print(f"Factorial = {fact}")
```

```python
#Reverse a Number
n = int(input("Enter a number: "))
a = 0
temp = n

while temp > 0:
  digit = temp % 10
  a = a * 10 + digit
  temp //= 10
print(a)
```

```python
#Palindorme Number

n = int(input("Enter a number: "))
temp = n
a = 0

while temp > 0 :
  digit = temp % 10
  a = a * 10 + digit
  temp //= 10

if a == temp:
  print("Number is plaindrome")
else:
  print("Number is not Planidrome")


```

```python
#Palindrome with for loop
word = input("Enter a word or number: ")
is_palindrome = True
length = len(word)

# We only need to check up to the middle of the word
for i in range(length // 2):
    if word[i] != word[length - 1 - i]:  # Compare front index with back index
        is_palindrome = False
        break

if is_palindrome:
    print("Yes, it is a Palindrome!")
else:
    print("No, it is not a Palindrome.")
```

```python
# Armstrong with while loop

n = int(input("enter a number:"))
temp = n
digits = 0

while temp> 0:
  temp//= 10
  digits = digits + 1

temp = n
sum_val = 0
while temp > 0:
  digit = temp % 10
  sum_val = sum_val + digit ** digits
  temp //= 10

if sum_val == n:
  print("Armstrong")
else:
  print("Not an Armstrong")
```
