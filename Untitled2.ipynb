```python
#Palindorme Number

n = int(input("Enter a number: "))
temp = n
a = 0

while temp > 0 :
  digit = temp % 10
  a = a * 10 + digit
  temp //= 10

if a == temp:
  print("Number is plaindrome")
else:
  print("Number is not Planidrome")


```

```python
#Palindrome with for loop
word = input("Enter a word or number: ")
is_palindrome = True
length = len(word)

# We only need to check up to the middle of the word
for i in range(length // 2):
    if word[i] != word[length - 1 - i]:  # Compare front index with back index
        is_palindrome = False
        break

if is_palindrome:
    print("Yes, it is a Palindrome!")
else:
    print("No, it is not a Palindrome.")
```

```python
# Armstrong with while loop

n = int(input("enter a number:"))
temp = n
digits = 0

while temp> 0:
  temp//= 10
  digits = digits + 1

temp = n
sum_val = 0
while temp > 0:
  digit = temp % 10
  sum_val = sum_val + digit ** digits
  temp //= 10

if sum_val == n:
  print("Armstrong")
else:
  print("Not an Armstrong")
```
